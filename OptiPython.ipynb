```python
#Tanque Cilindrico De Minimo Material
import numpy as np
from scipy.optimize import minimize

Vo= 0.5 #En m^3

#Funcion Objetivo

def area_sup(x):
  r,h = x[0],x[1]
  return 2*np.pi*r**2 +2*np.pi*r*h

#Restricciones
def restrict_volumen(x):
  r,h = x[0],x[1]
  return np.pi*r**2*h - Vo

#Config Problema
x0 = [0.3, 0.5]

#Restrict en scipy
restricciones = {
    'type': 'eq', #eq=equals
    'fun': restrict_volumen
}
#Limite: r&h>0
limite = [(0.01, None), (0.01, None)]

resultado = minimize(
    area_sup,
    x0,
    method='SLSQP',
    bounds=limite,
    constraints=restricciones)

r_opt, h_opt = resultado.x
print(" = " * 45)
print(" Tanque Cilindrico - Minimo material")
print(" = " * 45)

print(f"Estado: {'Optimo encontrado' if resultado.success else 'Fallo'}")
print()
print(f"Radio optimo : r = {r_opt:.2f} m  ({r_opt*100:.2f} cm)")

print(f"Altura optima : h = {h_opt:.2f} m ({h_opt*100:.2f} cm) ")

print(f"Area minima {resultado.fun:.4f} m^2")
print(f"Relacion h/r = {h_opt/r_opt:.2f} (teorico: 200)")

print(" = " * 45)

#Verificacion
vol_real=np.pi*r_opt**2*h_opt
print(f"Volumen Calculado {vol_real:8f} m^3 requerido : {Vo} m^3")
print(f"volumen en litros {vol_real*1000 :2f} L (requerido: 500 L)")
```

```python
# Mezcla de Fertilizantes - Maxima Ganancia
import numpy as np
from scipy.optimize import minimize

# Parametros
G1, G2 = 180, 220 # Ganancias por tonelada

# Funcion Objetivo
def ganancia_total(x):
  return -(G1*x[0] + G2*x[1]) # Negativo para maximizar

# Restricciones (Compuestos A, B, C)
def restrict_A(x): return 120 - (3*x[0] + 4*x[1])
def restrict_B(x): return 80 - (2*x[0] + 1*x[1])
def restrict_C(x): return 150 - (5*x[0] + 3*x[1])

# Config Problema
x0 = [1, 1]
restricciones = [
    {'type': 'ineq', 'fun': restrict_A},
    {'type': 'ineq', 'fun': restrict_B},
    {'type': 'ineq', 'fun': restrict_C}
]
limite = [(0, None), (0, None)]

resultado = minimize(
    ganancia_total,
    x0,
    method='SLSQP',
    bounds=limite,
    constraints=restricciones)

x1_opt, x2_opt = resultado.x
print(" = " * 45)
print(" Mezcla de Fertilizantes - Minimo Costo (Max Ganancia)")
print(" = " * 45)

print(f"Estado: {'Optimo encontrado' if resultado.success else 'Fallo'}")
print()
print(f"Produccion F1 : x1 = {x1_opt:.2f} ton")
print(f"Produccion F2 : x2 = {x2_opt:.2f} ton")

print(f"Ganancia maxima {-resultado.fun:.2f} $")
print(f"Uso de Compuesto A: {3*x1_opt + 4*x2_opt:.2f} kg / 120 kg")

print(" = " * 45)
```

```python

import numpy as np
from scipy.optimize import minimize

# Funcion Objetivo
def error_equilibrio(x):
  S, M = x[0], x[1]
  err_S = S - (0.6*S + 0.2*M + 40)
  err_M = M - (0.1*S + 0.5*M + 20)
  return err_S**2 + err_M**2 # El minimo debe ser 0

# Config Problema
x0 = [200, 80] # Valores iniciales
limite = [(0.01, None), (0.01, None)]

resultado = minimize(
    error_equilibrio,
    x0,
    method='SLSQP',
    bounds=limite)

S_eq, M_eq = resultado.x
print(" = " * 45)
print(" Produccion Semanal - Equilibrio (via Minimize)")
print(" = " * 45)

print(f"Estado: {'Equilibrio encontrado' if resultado.fun < 1e-6 else 'Fallo'}")
print()
print(f"Sillas en equilibrio: S = {S_eq:.2f} unidades")
print(f"Mesas en equilibrio : M = {M_eq:.2f} unidades")

print(f"Error residual del sistema: {resultado.fun:.8f}")
print(f"Verificacion Semana Sig: {0.6*S_eq + 0.2*M_eq + 40:.2f} (Debe ser igual a S)")

print(" = " * 45)
```

```python
# Epidemia SIS
import numpy as np
from scipy.optimize import minimize

# Parametros
N, beta, gamma = 10000, 0.00003, 0.1

# Funcion Objetivo
def error_derivada(I):
  dIdt = beta * (N - I) * I - gamma * I
  return dIdt**2 # El minimo es 0 en el equilibrio

# Config Problema
x0 = [5000] # Buscamos el equilibrio mayor a 0
limite = [(1, N)]

resultado = minimize(
    error_derivada,
    x0,
    method='SLSQP',
    bounds=limite)

I_opt = resultado.x[0]
print(" = " * 45)
print(" Modelo de Epidemia SIS - Punto Endemico")
print(" = " * 45)

print(f"Estado: {'Punto encontrado' if resultado.success else 'Fallo'}")
print()
print(f"Infectados estables: I = {I_opt:.2f} personas")
print(f"Poblacion sana : S = {N - I_opt:.2f} personas")

print(f"Valor de dI/dt en este punto: {beta*(N-I_opt)*I_opt - gamma*I_opt:.6f}")
print(f"Porcentaje de poblacion infectada: {(I_opt/N)*100:.2f} %")

print(" = " * 45)
```

```python
# Modelo Logistico de Pesca
import numpy as np
from scipy.optimize import minimize

# Parametros
r, K, H = 0.4, 1000, 80

# Funcion Objetivo
def error_crecimiento(P):
  crecimiento = r * P * (1 - P/K) - H
  return crecimiento**2 # El minimo es 0 en equilibrio

# Config Problema
x0_estable = [900] # Punto inicial cerca de la capacidad

resultado = minimize(
    error_crecimiento,
    x0_estable,
    method='SLSQP',
    bounds=[(0, K)])

P_opt = resultado.x[0]
print(" = " * 45)
print(" Modelo de Pesca - Equilibrio Estable")
print(" = " * 45)

print(f"Estado: {'Optimo encontrado' if resultado.fun < 1e-5 else 'Fallo'}")
print()
print(f"Poblacion estable : P = {P_opt:.2f} peces")
print(f"Poblacion inicial referenciada : 400 peces")

print(f"Crecimiento neto en equilibrio: {r*P_opt*(1-P_opt/K) - H:.6f}")
print(f"Poblacion sig (P_n+1): {P_opt + r*P_opt*(1-P_opt/K) - H:.2f} (Debe ser igual a P)")

print(" = " * 45)
```
